```python
from numpy.linalg import det

#Sample Input
num_var = int("2")
vars = "x y".split()
matrix = []
constants = []
coef_inputs = ["2 1", "1 4"]
const_inputs = ["4", "9"]
for i in range(num_var):
    matrix.append(list(map(float, coef_inputs[i].split())))
    constants.append(int(const_inputs[i]))

for i in range(num_var):
    temp_matrix = [row[:] for row in matrix]
    for j in range(num_var):
        temp_matrix[j][i] = constants[j]
    print(vars[i] + ": " + str(round(det(temp_matrix)/det(matrix), 3)))
```
